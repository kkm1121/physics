```python
import math

#변수 선언

g = 9.8; # 중력가속도 (m/s^2)
a = -g; # 가속도 = -중력가속도
v0 = 0; # 초기속도 (m/s)
PI = 3.141592; # 파이 = 3.141592;
```

```python
#초기 속도 입력
v0 = float(input())
print("초기속도 : ", v0,"m/s")
```

```python
#초기 각도
angle_degree = float(input())
print("초기 각도 : ", angle_degree,"도")

#각도를 radian으로 변환
angle_radian = angle_degree * PI / 180
print("초기 각도 : ", angle_radian,"radian")
```

```python
# x 성분의 속도를 구하세요
v0x = math.cos(angle_radian) * v0
print("x축의 초기속도 : ", v0x , "m/s")
```

```python
# y 성분의 속도를 구하세요
v0y = math.sin(angle_radian) * v0
print("y축의 초기 속도 : ", v0y, "m/s")
```

```python
# 최고점 높이

H = (v0y ** 2)/(2*g)
print("최고점 높이:",H,"m")
```

```python
# 최고점에 도착하는 시간을 구하여라
t = v0y / g
print("최고점 도달 시간:", t, "초")
```

```python
# 수평으로 이동한 최고 거리를 구하여라
x = v0x * t * 2
print("수평으로 이동한 최대거리:", x ,"m")
```

```python
# 지면 도착 시 속도 벡터(x,y, 전체 속도)
vx_final = v0x
vy_final = -v0y
v_final = math.sqrt(vx_final**2 + vy_final**2)
print("지면 도착 시 vx: ", vx_final,"m/s")
print("지면 도착시 vy: ", vy_final,"m/s")
print("지면 도착시 속도 벡터 크기 v:", v_final, "m/s")
```
